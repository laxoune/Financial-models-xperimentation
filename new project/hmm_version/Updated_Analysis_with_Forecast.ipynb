```python
# Data Loading and Preparation
# Ensure we are using NVDA stock data for the past 5 years
import yfinance as yf
import pandas as pd
import numpy as np

# Download NVDA stock data for the past 5 years
data = yf.download('NVDA', period='5y')
data['MA50'] = data['Close'].rolling(window=50).mean()
data['MA200'] = data['Close'].rolling(window=200).mean()
data['Daily Return'] = data['Close'].pct_change()
data.dropna(inplace=True)
```

```python
# Create features for supervised learning
# Ensure 'Target' column is not included as a feature
features = data[['Close', 'MA50', 'MA200', 'Daily Return']]
labels = data['Close'].shift(-1)  # Shift target for regression

# Drop NaN values after shift
features = features.iloc[:-1]
labels = labels.iloc[:-1]
```

```python
# Data Splitting Section
# Splitting the dataset into training and testing sets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.2, random_state=42, shuffle=False)
```

```python
# Feature Engineering Section
# Ensuring feature transformations are fit only on the training data and applied to both training and testing sets
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# LSTM Model Integration
from keras.models import Sequential
from keras.layers import LSTM, Dense, Dropout

# Prepare LSTM data (add time dimension)
X_train_lstm = np.expand_dims(X_train_scaled, axis=1)
X_test_lstm = np.expand_dims(X_test_scaled, axis=1)

# Build the LSTM Model
lstm_model = Sequential()
lstm_model.add(LSTM(units=50, return_sequences=True, input_shape=(X_train_lstm.shape[1], X_train_lstm.shape[2])))
lstm_model.add(Dropout(0.2))
lstm_model.add(LSTM(units=50, return_sequences=True))
lstm_model.add(Dropout(0.2))
lstm_model.add(LSTM(units=50))
lstm_model.add(Dropout(0.2))
lstm_model.add(Dense(units=1))

# Compile and train the LSTM model
lstm_model.compile(optimizer='adam', loss='mean_squared_error')
history = lstm_model.fit(X_train_lstm, y_train, epochs=100, batch_size=32, verbose=2)

# Plotting Training Loss
import matplotlib.pyplot as plt
plt.plot(history.history['loss'])
plt.title('Model Training Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.show()

# Predict using the LSTM model
lstm_train_pred = lstm_model.predict(X_train_lstm)
lstm_test_pred = lstm_model.predict(X_test_lstm)

# Add LSTM Predictions as a Feature
X_train_combined = np.hstack((X_train_scaled, lstm_train_pred))
X_test_combined = np.hstack((X_test_scaled, lstm_test_pred))

# Model Training Section (Updated for Stacking with LSTM)
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.linear_model import Ridge
from sklearn.ensemble import StackingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GridSearchCV

# StackingRegressor for Price Prediction
base_models = [
    ('rf', RandomForestRegressor(random_state=42)),
    ('svr', SVR(kernel='rbf')),
    ('ridge', Ridge())
]

stack_reg = StackingRegressor(
    estimators=base_models,
    final_estimator=LinearRegression()
)

param_grid = {
    'rf__n_estimators': [50, 100, 150],
    'rf__max_depth': [None, 10, 20],
    'svr__C': [0.1, 1, 10],
    'svr__gamma': ['scale', 'auto'],
    'ridge__alpha': [0.1, 1, 10]
}

grid_search = GridSearchCV(stack_reg, param_grid, cv=3, scoring='neg_mean_squared_error', verbose=2, n_jobs=-1)
grid_search.fit(X_train_combined, y_train)

# Best model after GridSearchCV
best_model = grid_search.best_estimator_
```

```python
# Model Evaluation Section
# Evaluate the stacking model on the test set and visualize the results
from sklearn.metrics import mean_squared_error, mean_absolute_error

stack_pred = best_model.predict(X_test_combined)
mse = mean_squared_error(y_test, stack_pred)
mae = mean_absolute_error(y_test, stack_pred)
rmse = np.sqrt(mse)
print(f"Stacking Model - MSE: {mse:.2f}, MAE: {mae:.2f}, RMSE: {rmse:.2f}")

# Save test evaluation results to a CSV file
evaluation_df = pd.DataFrame({
    'Actual Price': y_test.values.flatten(),
    'Predicted Price': stack_pred.flatten()
}, index=y_test.index)
evaluation_df.to_csv('test_evaluation.csv', index_label='Date')

# Evaluate Trend Accuracy
# Trend accuracy measures the percentage of correct predictions for price direction
actual_direction = np.sign(y_test.values[1:] - y_test.values[:-1])  # 1 if price increases, -1 if decreases
predicted_direction = np.sign(stack_pred[1:] - stack_pred[:-1])
trend_accuracy = (actual_direction == predicted_direction).mean() * 100
print(f"Trend Accuracy: {trend_accuracy:.2f}%")

# Visualization of test set predictions
plt.figure(figsize=(14, 7))
plt.plot(y_test.index, y_test.values, label='Actual Prices', color='blue')
plt.plot(y_test.index, stack_pred, label='Stacking Prediction', color='orange')
plt.title('Stock Price Prediction vs Actual (Stacking + LSTM Model)')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Forecasting Section
# Generate a 20 trading day forecast

# Assume the last 20 days of test set features are used for prediction
forecast_features = X_test_combined[-20:]
forecast = best_model.predict(forecast_features)

# Create a DataFrame for the forecast
forecast_df = pd.DataFrame({
    'Date': data.index[-20:],
    'Predicted Price': forecast.flatten()
})

# Save the forecast to a CSV file
forecast_df.to_csv('20_day_forecast.csv', index=False)

# Visualization of the 20-day forecast
plt.figure(figsize=(14, 7))
plt.plot(forecast_df['Date'], forecast_df['Predicted Price'], label='20-Day Forecast', marker='o')
plt.title('20-Day Stock Price Forecast')
plt.xlabel('Date')
plt.ylabel('Predicted Price')
plt.legend()
plt.show()
```
